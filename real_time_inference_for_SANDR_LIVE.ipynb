```python
#MULTI-MODEL ENGINE GENERATOR
import nest_asyncio
nest_asyncio.apply()
import asyncio
import datetime as dt
import pandas as pd
import numpy as np
import joblib
import re
import time
from collections import deque
import os
import json
from datetime import datetime
import os
from dotenv import load_dotenv

# 1. Load the variables from the .env file you just created
load_dotenv(dotenv_path="live.env", override=True)
# ==========================================
# 1. THE "SMART" REAL-TIME ENGINE (From Block 1)
# ==========================================
class TradingModelRealTime:
    def __init__(self, model_pkl_path, static_features_csv_path, buffer_size=100):
        # Load Model Data
        self.data = joblib.load(model_pkl_path)
        self.model = self.data['model']
        self.feature_order = self.data['features']
        
        # Try to get symbol from pickle, otherwise infer from filename
        if 'symbol' in self.data:
            self.symbol = self.data['symbol']
        else:
            self.symbol = os.path.basename(model_pkl_path).split('_')[2]
        
        # --- State Buffers ---
        # Stores (timestamp, price) for RSI/Volatility calcs
        self.price_history = {} 
        # Stores (timestamp, size) for Volume calcs
        self.volume_history = {} 
        self.buffer_size = buffer_size

        # Load Static Baseline (The big CSV)
        full_df = static_features_csv_path#pd.read_csv(static_features_csv_path)
        self.baseline_row = full_df.iloc[-1].to_dict()
        
        # Build the Recipe Map (The "Smart" Logic)
        self.feature_map = self._build_feature_recipes()

    
            
    def _build_feature_recipes(self):
        recipes = {}
        patterns = {
            'cl': r"change_lag_(\d+)(\d{2})th_ptile",
            'zs': r"z_score_(\d+)?",
            'rv': r"relative_volume_(\d+)",
            'vf': r"volume_force_(\d+)",
            'rsi': r"rsi_(\d+)",
            'sarv': r"session_adj_rel_volume_(\d+)"
        }
        
        # Mapping feature string keywords to your internal interaction types
        interaction_keywords = {
            'Open30m': 'is_open_30m',
            'Close30m': 'is_close_30m',
            'MidDay': 'is_mid_day',
            'Mon_Interact': 'is_mon',
            'Fri_Interact': 'is_fri',
            'earn_prox': 'earn_prox'
        }

        for feat in self.feature_order:
            ticker = feat.split('_')[-1]
            recipe = {'name': feat, 'is_static': True, 'interaction': None, 'ticker': ticker}
            
            # Detect Interaction from string
            for kw, i_type in interaction_keywords.items():
                if kw in feat:
                    recipe['interaction'] = i_type
                    break

            for p_type, regex in patterns.items():
                m = re.search(regex, feat)
                if m:
                    recipe['is_static'] = False
                    recipe['type'] = p_type
                    recipe['ticker'] = ticker # We ignore interactions for now as they are baked into CSV
                    
                    groups = m.groups()
                    lag = groups[0]
                    
                    if p_type == 'cl':
                        ptile = groups[1]
                        recipe.update({'lag': lag, 'ptile': ptile})
                        recipe['ancillary'] = [f'{ptile}th_ptile_lag_{lag}days_{ticker}']
                        
                    elif p_type == 'zs':
                        recipe.update({'lag': lag})
                        recipe['ancillary'] = [f'rolling_mean_{lag}_{ticker}', f'rolling_std_{lag}_{ticker}']
                        
                    elif p_type in ['rv', 'sarv']:
                        recipe.update({'lag': lag})
                        recipe['ancillary'] = [f'rolling_avg_vol_{lag}_{ticker}']
                        
                    elif p_type == 'vf':
                        recipe.update({'lag': lag})
                        recipe['ancillary'] = [f'rolling_avg_vol_{lag}_{ticker}', f'mean_vwap_15_{ticker}'] 
                        
                    elif p_type == 'rsi':
                        recipe.update({'lag': lag})
                    
                    matched = True
                    break
            
            recipes[feat] = recipe
            
        return recipes

    def update_market_data(self, symbol, price, size=0):
        """Updates internal buffers with live tick data."""
        now = time.time()
        
        # Init buffers if new symbol
        if symbol not in self.price_history:
            self.price_history[symbol] = deque(maxlen=self.buffer_size)
            self.volume_history[symbol] = deque() 
            
        # Update Price
        self.price_history[symbol].append(price)
        
        # Update Volume (with timestamp for 15m sliding window)
        if size > 0:
            self.volume_history[symbol].append((now, size))
            
        # Clean old volume trades (> 15 mins)
        while self.volume_history[symbol] and self.volume_history[symbol][0][0] < now - 900:
            self.volume_history[symbol].popleft()

    def _get_current_volume(self, ticker):
        """Sum of volume in the last 15 minutes."""
        if ticker in self.volume_history:
            return sum(t[1] for t in self.volume_history[ticker])
        return 0

    def _calc_live_rsi(self, ticker, period=14):
        """Approximate live RSI from price buffer."""
        if ticker not in self.price_history or len(self.price_history[ticker]) < period + 1:
            return 50 # Default neutral
            
        prices = list(self.price_history[ticker])
        series = pd.Series(prices)
        delta = series.diff()
        gain = (delta.where(delta > 0, 0)).ewm(span=period).mean()
        loss = (-delta.where(delta < 0, 0)).ewm(span=period).mean()
        
        rs = gain.iloc[-1] / loss.iloc[-1] if loss.iloc[-1] != 0 else 0
        return 100 - (100 / (1 + rs))

    def predict(self, live_prices):
        # We need SPY for relative calculations
        spy_price = live_prices.get('SPY')
        if not spy_price: return None
        
        # 1. Calculate Global Time Vars
        now_dt = dt.datetime.now()
        minutes_since_open = (now_dt.hour - 9) * 60 + (now_dt.minute - 30)
        day_of_week = now_dt.weekday() # 0=Mon, 4=Fri
        
        # 2. Build Input Vector
        input_vector = {}
        
        for feat, recipe in self.feature_map.items():
            val = 0
            ticker = recipe['ticker']
            # --- STEP 1: GET BASE VALUE ---
            if recipe['is_static']:
                val = self.baseline_row.get(feat, 0)
            else:
                if ticker not in live_prices:
                    input_vector[feat] = 0
                    continue
                    
                curr_price = live_prices[ticker]
                rel_price = curr_price / spy_price
                
                if recipe['type'] == 'cl':
                    base_lag_val = self.baseline_row.get(recipe['ancillary'][0], 1)
                    val = (rel_price - base_lag_val) / base_lag_val 
                elif recipe['type'] == 'zs':
                    mu = self.baseline_row.get(recipe['ancillary'][0], 0)
                    sigma = self.baseline_row.get(recipe['ancillary'][1], 1)
                    val = (rel_price - mu) / sigma if sigma != 0 else 0
                elif recipe['type'] in ['rv', 'sarv']:
                    avg_vol = self.baseline_row.get(recipe['ancillary'][0], 1)
                    val = self._get_current_volume(ticker) / avg_vol
                    if recipe['type'] == 'sarv':
                        # Session adjust (U-shape volume curve)
                        progress = max(0, min(1, minutes_since_open / 390.0))
                        adj = 1 - abs(2 * progress - 1) + 0.5
                        val *= adj
                elif recipe['type'] == 'vf':
                    prev_rel_vwap = self.baseline_row.get(recipe['ancillary'][1], rel_price) 
                    pct_chg = (rel_price / prev_rel_vwap) - 1 if prev_rel_vwap != 0 else 0
                    rel_vol = self._get_current_volume(ticker) / self.baseline_row.get(recipe['ancillary'][0], 1)
                    val = pct_chg * rel_vol
                elif recipe['type'] == 'rsi':
                    val = self._calc_live_rsi(ticker)

            # --- STEP 2: APPLY INTERACTIONS (To Static OR Dynamic) ---
            if recipe['interaction']:
                mask = 0 # Default to 0 for binary masks
                i_type = recipe['interaction']
                
                if i_type == 'is_open_30m':
                    mask = 1 if 0 <= minutes_since_open <= 30 else 0
                elif i_type == 'is_close_30m':
                    mask = 1 if minutes_since_open >= 360 else 0
                elif i_type == 'is_mid_day':
                    mask = 1 if 30 < minutes_since_open < 360 else 0
                elif i_type == 'is_mon':
                    mask = 1 if day_of_week == 0 else 0
                elif i_type == 'is_fri':
                    mask = 1 if day_of_week == 4 else 0
                elif i_type == 'earn_prox':
                    days = self.baseline_row.get(f'days_before_earnings_{ticker}', 90)
                    mask = 1 / (days + 1)
                
                val *= mask
                
            input_vector[feat] = val

        # 3. Inference
        X = pd.DataFrame([input_vector])[self.feature_order]
        return self.model.predict_proba(X)[0][1]
            

# ==========================================
# 2. GLOBAL CONFIG & MULTI-MODEL SETUP (From Block 2)
# ==========================================
#API_KEY = "AKH3ZFJ2LWAKH7SSV2AHKD4PK4"
#SECRET_KEY = "7gGJDh5UAxh3gg8V21d3AcuPfNJcwtKqukiqMhdhsjrW"

API_KEY = os.getenv("ALPACA_API_KEY")
SECRET_KEY = os.getenv("ALPACA_SECRET_KEY")
import glob

print(f"✅ Loaded {len(VALID_MODEL_PATHS)} models from directory.")


print(f"\nFinal Live Stream List ({len(LIVE_WATCHLIST)}): {LIVE_WATCHLIST}")


CSV_PATH=pd.read_csv('daily_feature_pres/feature_presentation_'+industry+'.csv')
# ==========================================
# 4. INITIALIZE ENGINE FLEET
# ==========================================
# Now we initialize the SMART engines for every model in our list
engines = [TradingModelRealTime(path, CSV_PATH) for path in VALID_MODEL_PATHS]

print(f"Successfully initialized {len(engines)} smart model engines.")
print("Ready for Real-Time Loop.")
```

```python
import pytz
import asyncio
import datetime as dt
import time
from alpaca.data.live.stock import StockDataStream
from alpaca.trading.requests import MarketOrderRequest
from alpaca.trading.enums import OrderSide, TimeInForce
from alpaca.trading.client import TradingClient

# ==========================================
# GLOBAL STATE & CONFIG
# ==========================================
live_prices = {}
PREDICTION_INTERVAL = 60 
HEARTBEAT_INTERVAL = 900 
MAX_PORTFOLIO_VALUE = 10.00

# EXIT CUTOFFS & TIMEOUTS
TAKE_PROFIT_PCT = 0.045   # +4.5%
STOP_LOSS_PCT = -0.05     # -5.0%
DECAY_THRESHOLD = 0.60
COOLDOWN_MINUTES = 30     # Standard cooldown after a buy
STOP_PENALTY_MINUTES = 120 # Timeout after a stop-loss hit

# --- TRADING CONFIG ---
bet = 10000 # Standard order size
cooldown_tracker = {} # Stores {symbol: timestamp_when_cooldown_expires}

# Ensure 'trading_client' is defined (from your previous cells)
trading_client = TradingClient(API_KEY, SECRET_KEY, paper=False)

# ==========================================
# 3. ORDER EXECUTION LOGIC
# ==========================================

async def execute_trade(symbol, probability, price, trade_side='LONG', custom_amount=None):
    """
    Places a buy (long) or sell (short) order based on the model's prediction.
    Accepts a 'custom_amount' to allow for dynamic hedge sizing.
    """
    now = time.time()
    trade_amount = custom_amount if custom_amount else bet

    account = trading_client.get_account()
    buying_power = float(account.non_marginable_buying_power) 
    # 1. Check Cooldown (Now tracking expiration time instead of start time)
    if symbol in cooldown_tracker and now < cooldown_tracker[symbol]:
        mins_left = int((cooldown_tracker[symbol] - now) / 60)
        print(f"⏸️ [SKIP] {symbol} is in cooldown/penalty ({mins_left}m remaining).")
        return

    try:
        # 2. Check Portfolio Limit
        current_holdings = float(account.long_market_value) + abs(float(account.short_market_value))
        
        if (current_holdings + trade_amount) > MAX_PORTFOLIO_VALUE:
            print(f"🛑 [MAX REACHED] Skipping {symbol}. Adding ${trade_amount:,.2f} exceeds $50k limit.")
            return
        # Inside execute_trade...
        

        if trade_amount > buying_power:
            print(f"❌ [INSUFFICIENT FUNDS] Need ${trade_amount}, only have ${buying_power}")
            return

        # 3. Prepare the Order Details based on Trade Side
        if trade_side == 'SHORT':
            if price <= 0:
                print(f"❌ [ERROR] Invalid price (${price}) for {symbol}. Cannot calculate short qty.")
                return
                
            qty_to_short = int(round(trade_amount / price, 0))
            qty_to_short = max(1, qty_to_short) # Ensure we short at least 1 share
            
            print(f"🩸 [EXECUTE] SHORTING {qty_to_short} shares of {symbol} (Prob: {probability:.4f}, Est. ${trade_amount:,.2f})")
            
            order_data = MarketOrderRequest(
                symbol=symbol,
                qty=qty_to_short,
                side=OrderSide.SELL,
                time_in_force=TimeInForce.DAY
            )
            
        else: # LONG
            print(f"🚀 [EXECUTE] LONGING ${trade_amount:,.2f} of {symbol} (Prob: {probability:.4f})")
            
            order_data = MarketOrderRequest(
                symbol=symbol,
                notional=trade_amount,
                side=OrderSide.BUY,
                time_in_force=TimeInForce.DAY
            )

        # Submit to Alpaca
        trade = trading_client.submit_order(order_data)
        
        print(f"✅ [FILLED] {trade_side} Order ID: {trade.id} | Status: {trade.status}")
        
        # Set Standard Cooldown expiration
        cooldown_tracker[symbol] = now + (COOLDOWN_MINUTES * 60)
        
    except Exception as e:
        print(f"❌ [ORDER FAILED] Could not place {trade_side} trade for {symbol}: {e}")

# ==========================================
# 4. ASYNC TASKS
# ==========================================

async def on_trade_update(data):
    for engine in engines:
        if engine.symbol == data.symbol:
            engine.update_market_data(data.symbol, data.price, size=data.size)

async def on_quote_update(data):
    if data.bid_price > 0 and data.ask_price > 0:
        mid_price = (data.bid_price + data.ask_price) / 2
        live_prices[data.symbol] = mid_price

async def inference_loop():
    print(f"[INFO] Inference loop started (Interval: {PREDICTION_INTERVAL}s)")
    last_heartbeat = time.time()
    
    while True:
        try:
            await asyncio.sleep(PREDICTION_INTERVAL)
            
            now_ny = dt.datetime.now(pytz.timezone('America/New_York'))
            timestamp_str = now_ny.strftime('%H:%M:%S')

            # --- 9:31 AM MARKET START DELAY ---
            if now_ny.time() < dt.time(9, 31):
                print(f"[WAIT] Waiting for market open (9:31 AM ET). Current: {timestamp_str}")
                continue
            
            if 'SPY' not in live_prices:
                print(f"[WAIT] Waiting for SPY data... (Got {len(live_prices)} prices)")
                continue

            # ==========================================
            # STEP A: EVALUATE EXITS (HYBRID LOGIC)
            # ==========================================
            try:
                positions = trading_client.get_all_positions()
                for pos in positions:
                    sym = pos.symbol
                    unrealized_pct = float(pos.unrealized_plpc)
                    side_str = str(pos.side).lower() # Returns 'long' or 'short'
                    
                    close_reason = None
                    is_stop_loss = False
                    
                    # 1. HARD STOP-LOSS & TAKE-PROFIT
                    if unrealized_pct >= TAKE_PROFIT_PCT:
                        close_reason = f"TAKE PROFIT (+{unrealized_pct*100:.2f}%)"
                    elif unrealized_pct <= STOP_LOSS_PCT:
                        close_reason = f"STOP LOSS ({unrealized_pct*100:.2f}%)"
                        is_stop_loss = True
                    else:
                        # 2. MODEL DECAY TRIGGER
                        for engine in engines:
                            if engine.symbol == sym:
                                model_id = getattr(engine, 'name', str(engine))
                                is_down_model = '_15' in model_id
                                
                                if (side_str == 'short' and is_down_model) or (side_str == 'long' and not is_down_model):
                                    prob = engine.predict(live_prices)
                                    if prob is not None and prob < DECAY_THRESHOLD:
                                        close_reason = f"MODEL DECAY (Prob dropped to {prob:.4f})"
                                break 
                    
                    # EXECUTE CLOSE
                    if close_reason:
                        print(f"🛑 [CLOSING POSITION] {sym} | Reason: {close_reason}")
                        trading_client.close_position(sym)
                        
                        # --- PENALTY BOX LOGIC ---
                        if is_stop_loss:
                            # 2 Hour penalty before it can be bought again
                            cooldown_tracker[sym] = time.time() + (STOP_PENALTY_MINUTES * 60)
                            print(f"⏱️ [PENALTY] {sym} sent to penalty box for {STOP_PENALTY_MINUTES} mins.")
                        else:
                            # Natural exit, clear cooldown so it can trigger immediately on next setup
                            if sym in cooldown_tracker:
                                del cooldown_tracker[sym]
                            
            except Exception as e:
                print(f"❌ [EXIT EVALUATION ERROR]: {e}")

            # ==========================================
            # STEP B: EVALUATE ENTRIES (RUN INFERENCE)
            # ==========================================
            high_conviction = []
            all_probs = []
            
            for engine in engines:
                prob = engine.predict(live_prices)
                
                if prob is not None:
                    model_identifier = getattr(engine, 'name', str(engine)) 
                    is_short = '_15' in model_identifier
                    all_probs.append((f"{engine.symbol} ({'SHORT' if is_short else 'LONG'})", prob, engine))
                    
                    # === TRADING TRIGGER ===
                    if prob > 0.85:
                        current_price = live_prices.get(engine.symbol, 0)
                        
                        already_held = any(p.symbol == engine.symbol for p in positions)
                        if already_held:
                            continue
                            
                        if is_short:
                            high_conviction.append((f"{engine.symbol} (SHORT)", prob))
                            print(f"⚠️ [SHORT ALERT] {engine.symbol} DOWN-MOVE CONVICTION: {prob:.4f} at {timestamp_str}")
                            
                            asyncio.create_task(execute_trade(
                                symbol=engine.symbol, probability=prob, price=current_price, trade_side='SHORT'
                            ))
                        else:
                            high_conviction.append((f"{engine.symbol} (LONG)", prob))
                            print(f"🚀 [LONG ALERT] {engine.symbol} UP-MOVE CONVICTION: {prob:.4f} at {timestamp_str}")
                            
                            asyncio.create_task(execute_trade(
                                symbol=engine.symbol, probability=prob, price=current_price, trade_side='LONG'
                            ))

            # ==========================================
            # STEP C: MARKET NEUTRALITY BALANCING
            # ==========================================
            try:
                account = trading_client.get_account()
                long_val = float(account.long_market_value)
                short_val = abs(float(account.short_market_value))
                total_val = long_val + short_val
                
                if total_val > 0:
                    diff_pct = abs(long_val - short_val) / total_val
                    
                    if diff_pct > 0.025:
                        # Find exactly how much money we need to balance the scales perfectly
                        diff_amount = abs(long_val - short_val)
                        
                        positions = trading_client.get_all_positions()
                        held_symbols = [p.symbol for p in positions]
                        
                        # Sort all evaluated models from highest prob to lowest
                        all_probs.sort(key=lambda x: x[1], reverse=True)
                        
                        if long_val > short_val:
                            # Too long -> Need to find the best SHORT to hedge
                            for label, prob, eng in all_probs:
                                if 'SHORT' in label and eng.symbol not in held_symbols:
                                    print(f"⚖️ [REBALANCE] Hedging ${diff_amount:,.2f} with {eng.symbol} SHORT (Prob: {prob:.4f})")
                                    current_price = live_prices.get(eng.symbol, 0)
                                    asyncio.create_task(execute_trade(
                                        symbol=eng.symbol, probability=prob, price=current_price, 
                                        trade_side='SHORT', custom_amount=diff_amount
                                    ))
                                    break 
                                    
                        elif short_val > long_val:
                            # Too short -> Need to find the best LONG to hedge
                            for label, prob, eng in all_probs:
                                if 'LONG' in label and eng.symbol not in held_symbols:
                                    print(f"⚖️ [REBALANCE] Hedging ${diff_amount:,.2f} with {eng.symbol} LONG (Prob: {prob:.4f})")
                                    current_price = live_prices.get(eng.symbol, 0)
                                    asyncio.create_task(execute_trade(
                                        symbol=eng.symbol, probability=prob, price=current_price, 
                                        trade_side='LONG', custom_amount=diff_amount
                                    ))
                                    break
                                    
            except Exception as e:
                print(f"❌ [NEUTRALITY ERROR]: {e}")

            # --- HEARTBEAT LOGGING ---
            if (time.time() - last_heartbeat > HEARTBEAT_INTERVAL) or len(high_conviction) > 0:
                print(f"\n[{timestamp_str}] === MARKET HEARTBEAT ===")
                all_probs.sort(key=lambda x: x[1], reverse=True)
                top_3 = all_probs[:3]
                print(f"Top Models: {', '.join([f'{label}: {p:.4f}' for label, p, eng in top_3])}")
                
                sample_tickers = list(live_prices.keys())[:5]
                print(f"Live Prices ({len(live_prices)} total): { {k: round(live_prices[k], 2) for k in sample_tickers} }")
                print("==================================\n")
                last_heartbeat = time.time()
                
        except Exception as e:
            print(f"[ERROR] Inference loop crashed: {e}")
            await asyncio.sleep(1)

async def main():
    ny_time = dt.datetime.now(pytz.timezone('America/New_York')).time()
    if ny_time >= dt.time(16, 0): 
        print("[STOP] Market is closed (After 16:00 ET).")
        return

    while True:
        inference_task = None
        try:
            print(f"[START] Connecting to Alpaca Stream... {dt.datetime.now()}")
            stream = StockDataStream(API_KEY, SECRET_KEY)
            stream.subscribe_trades(on_trade_update, *LIVE_WATCHLIST)
            stream.subscribe_quotes(on_quote_update, *LIVE_WATCHLIST)

            inference_task = asyncio.create_task(inference_loop())
            await stream.run()

        except Exception as e:
            print(f"[RECONNECT] Stream connection lost: {e}")
            if inference_task: inference_task.cancel()
            print("Retrying in 5 seconds...")
            await asyncio.sleep(5)

# --- EXECUTION ---
await main()
```
